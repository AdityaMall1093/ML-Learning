```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('train.txt',sep=';',header=None,names=['text','emotion'])
```

```python
unique_emotion=df['emotion'].unique()
emotion_number={}
i=0
for emo in unique_emotion:
    emotion_number[emo]=i
    i +=1
df['emotion']=df['emotion'].map(emotion_number)
```

```python
df
```

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

```python
import string

def remove_punc(text):
    if isinstance(text, str):
        return text.translate(str.maketrans('', '', string.punctuation))
    return text

df['text'] = df['text'].apply(remove_punc)
```

```python
def remove_number(txt):
    new = ""

    for i in txt:
        if not i.isdigit():
            new = new + i

    return new

df['text'] = df['text'].apply(remove_number)
```

```python
def remove_emojis(txt):
   new=""

   for i in txt:
    if i.isascii():
       new=new+i

   return new

df['text']=df['text'].apply(remove_emojis)

```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
len(stop_words)
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words=txt.split()
  cleaned=[]
  for i in words:
    if not i in stop_words:
      cleaned.append(i)

  return ''.join(cleaned)
```

```python
df['text']=df['text'].apply(remove)
```

```python
df.loc[1]['text']
```

```python
from re import X
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.2, random_state=42)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer
```

```python
bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
nb_classifier = MultinomialNB()
nb_classifier.fit(X_train_bow, y_train)
```

```python
pred_bow = nb_classifier.predict(X_test_bow)
print(accuracy_score(y_test, pred_bow))
print(classification_report(y_test, pred_bow))
```

```python
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)
```

```python
nb2_model=MultinomialNB()
nb2_model.fit(X_train_tfidf,y_train)
```

```python
y_pred=nb2_model.predict(X_test_tfidf)
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_reg=LogisticRegression(max_iter=1000)
log_reg.fit(X_train_tfidf,y_train)
```

```python
log_pred=log_reg.predict(X_test_tfidf)
print(accuracy_score(y_test,log_pred))
print(classification_report(y_test,log_pred))
```
